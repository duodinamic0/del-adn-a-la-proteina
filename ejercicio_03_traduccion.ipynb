# Ejercicio 3: Traducción del ARNm a Proteína y Extensión con Biopython

## 📖 Introducción y Fundamento Teórico Resumido

> **Objetivo:** Aplicar el código genético universal para traducir con precisión una pauta de lectura de ARNm a una secuencia polipeptídica funcional, identificar los codones de inicio y parada, y reflexionar sobre el impacto biológico de mutaciones críticas.

### Resumen de los puntos teóricos del enunciado:

1. **Transcrito de ARNm de partida (5' -> 3'):**
   $$\mathbf{5'\text{ – AUG UAU GCU UAA – }3'}$$
   Consta de 12 ribonucleótidos organizados en 4 tripletes o codones continuos y no solapantes.

2. **Identificación de codones reguladores:**
   - **Codón de Inicio:** $\mathbf{5'\text{-AUG-}3'}$ (Codón 1). Establece el marco de lectura e incorpora **Metionina (Met / M)** (o N-formilmetionina en bacterias).
   - **Codón de Paro / Terminación:** $\mathbf{5'\text{-UAA-}3'}$ (Codón 4, *Ochre*). No codifica ningún aminoácido; es reconocido por factores de liberación proteicos (RF) que liberan la cadena polipeptídica y disocian el ribosoma.

3. **Traducción manual al péptido funcional:**
   - **AUG** $\implies$ Metionina (Met / M)
   - **UAU** $\implies$ Tirosina (Tyr / Y)
   - **GCU** $\implies$ Alanina (Ala / A)
   - **UAA** $\implies$ Stop (*)
   $$\text{Péptido maduro resultante:} \quad \mathbf{\text{Met – Tyr – Ala}} \quad (\mathbf{\text{MYA}})$$

4. **Reflexión sobre mutaciones críticas:**
   - **Mutación en el codón de inicio ($AUG \to GUG$):**
     - En eucariotas el ribosoma necesita estrictamente $AUG$ (consenso Kozak); la sustitución por $GUG$ reduce la iniciación $>95\%$, aboliendo la expresión del gen o forzando inicios aberrantes *downstream* fuera de fase.
     - En bacterias $GUG$ puede actuar como inicio alternativo funcional con fMet, pero con menor eficiencia ($10-30\%$).
   - **Pérdida del codón de paro (*Stop-loss*, ej. $UAA \to CAA$):**
     - El ribosoma no se detiene y continúa traduciendo hacia la región $3'\text{-UTR}$ (*translational readthrough*).
     - Genera proteínas anómalas con extensiones C-terminales patogénicas, agregación celular o estancamiento del ribosoma al llegar a la cola poli(A), activando rutas de degradación como *Non-stop Decay* (NSD).

# 💻 Extensión con Biopython: Guía Paso a Paso

> **Enunciado de la extensión:**
> *Utiliza el módulo Bio.Seq para traducir automáticamente el ARNm a una secuencia proteica y comprueba si el resultado coincide con el tuyo.*

A continuación, implementaremos el script de traducción y simularemos computacionalmente las mutaciones planteadas de forma guiada e interactiva.

### Paso 1: Importar la clase `Seq` y definir el transcrito de ARNm

Creamos el objeto `Seq` con la secuencia `AUGUAUGCUUAA` y exploramos su pauta de lectura desglosando los 4 tripletes (codones).

```python
from Bio.Seq import Seq

# 1. Definición del transcrito de ARNm dado (5' -> 3')
mrna_str = "AUGUAUGCUUAA"
mrna_seq = Seq(mrna_str)

print("=" * 65)
print("EJERCICIO 3: TRADUCCIÓN DEL ARNm A PROTEÍNA CON BIOPYTHON")
print("=" * 65)
print(f"Transcrito de ARNm (5' -> 3'): {mrna_seq}")
print(f"Longitud total:                {len(mrna_seq)} ribonucleótidos")

# Desglose en codones de 3 nucleótidos
codones = [mrna_str[i:i+3] for i in range(0, len(mrna_str), 3)]
print(f"Codones identificados:         {codones}")
print(f"  - Codón 1 (Inicio):          {codones[0]} -> Metionina (M)")
print(f"  - Codones 2 y 3 (Cuerpo):    {codones[1:-1]}")
print(f"  - Codón 4 (Parada / Stop):   {codones[-1]} -> Señal de parada (*)")
print("=" * 65)
```

### Paso 2: Traducción con Biopython (`.translate()`)

Biopython incorpora el método biológico **`.translate()`**, el cual consulta la tabla del código genético universal estándar.

Es crucial distinguir dos modalidades del método:
1. **`mrna_seq.translate()` (por defecto):** Traduce todos los codones y representa el codón de parada con el símbolo asterisco `*`.
2. **`mrna_seq.translate(to_stop=True)`:** Detiene la traducción exactamente al encontrar el primer codón de parada, devolviendo únicamente la secuencia de aminoácidos del péptido funcional final.

```python
# 2. Traducción automática con Biopython

# Modalidad A: incluye el asterisco '*' para el codón de parada
protein_with_stop = mrna_seq.translate()

# Modalidad B: trunca al llegar al codón de parada (péptido funcional)
protein_to_stop = mrna_seq.translate(to_stop=True)

print("=" * 65)
print("RESULTADOS DE LA TRADUCCIÓN CON BIOPYTHON")
print("=" * 65)
print(f"Secuencia con símbolo Stop ('*'):  {protein_with_stop}")
print(f"Péptido maduro funcional:          {protein_to_stop}")
print(f"Longitud del péptido funcional:    {len(protein_to_stop)} aminoácidos")
print("=" * 65)
```

### Paso 3: Comprobación automatizada frente al resultado manual

El enunciado indica: *«comprueba si el resultado coincide con el tuyo»*.

En la resolución teórica manual determinamos:
- Péptido con parada: `MYA*`
- Péptido funcional maduro: `MYA` (Metionina - Tirosina - Alanina)

Verificamos la coincidencia exacta mediante aserciones (`assert`):

```python
# 3. Comprobación contra la resolución manual teórica
manual_protein_with_stop = "MYA*"
manual_peptide = "MYA"

assert str(protein_with_stop) == manual_protein_with_stop, (
    f"Discrepancia con stop: esperado {manual_protein_with_stop}, obtenido {protein_with_stop}"
)
assert str(protein_to_stop) == manual_peptide, (
    f"Discrepancia sin stop: esperado {manual_peptide}, obtenido {protein_to_stop}"
)

print("=" * 65)
print(">> VERIFICACIÓN MANUAL vs BIOPYTHON: EXITOSA (COINCIDENCIA EXACTA)")
print(f"   [OK] Con stop:    {protein_with_stop} == {manual_protein_with_stop}")
print(f"   [OK] Péptido:     {protein_to_stop} == {manual_peptide}")
print("   Residuos: Metionina (M) - Tirosina (Y) - Alanina (A)")
print("=" * 65)
```

### Paso 4: Simulación de la Mutación en el Codón de Inicio (AUG -> GUG)

Vamos a simular la primera pregunta biológica del enunciado: *¿qué pasaría si el codón de inicio mutara de AUG a GUG?*

Generamos la secuencia mutada `GUGUAUGCUUAA` y la traducimos para contrastar la salida estándar algorítmica frente a la realidad celular.

```python
# 4.1. Mutación puntual en el codón de inicio: AUG -> GUG
mut_inicio_str = "GUGUAUGCUUAA"
mut_inicio_seq = Seq(mut_inicio_str)

# Traducción con código genético estándar
prot_mut_inicio = mut_inicio_seq.translate(to_stop=True)

print("=" * 65)
print("SIMULACIÓN 1: MUTACIÓN EN EL CODÓN DE INICIO (AUG -> GUG)")
print("=" * 65)
print(f"ARNm normal:                    5'- {mrna_seq} -3'")
print(f"ARNm mutado:                    5'- {mut_inicio_seq} -3'")
print(f"Péptido en código estándar:     {prot_mut_inicio} (Val - Tyr - Ala)")

print("\n>> INTERPRETACIÓN BIOLÓGICA:")
print("1. En EUCARIOTAS: La traducción se pierde por completo (>95% de inhibición),")
print("   ya que el complejo de escaneo 43S requiere estrictamente el triplete AUG")
print("   complementario al Met-tRNA_i. Puede forzar inicios aberrantes downstream.")
print("2. En PROCARIOTAS: GUG puede actuar como inicio alternativo funcional con fMet,")
print("   pero con una eficiencia sustancialmente reducida (10-30% respecto a AUG).")
print("=" * 65)
```

### Paso 5: Simulación de Pérdida de Parada (*Stop-Loss*, UAA -> CAA)

Ahora simulamos la segunda pregunta: *¿qué ocurriría si el codón de paro desapareciera por mutación?*

Si el codón de paro `UAA` muta a `CAA` (Glutamina / Q), el ribosoma continúa leyendo a través de la región no traducida $3'\text{-UTR}$ (*untranslated region*) hasta encontrar un codón de parada críptico secundario (o la cola de poli-A).

```python
# 4.2. Mutación Stop-loss: UAA muta a CAA y se prolonga en la región 3'-UTR
# Añadimos una región 3'-UTR hipotética que contiene un codón de parada críptico (UGA):
mut_stoploss_str = "AUGUAUGCUCAAGCCUGA"  # UAA mutó a CAA (Gln), sigue GCC (Ala) y STOP UGA
mut_stoploss_seq = Seq(mut_stoploss_str)
prot_mut_stoploss = mut_stoploss_seq.translate(to_stop=True)

print("=" * 65)
print("SIMULACIÓN 2: MUTACIÓN DE PÉRDIDA DE PARADA (STOP-LOSS, UAA -> CAA)")
print("=" * 65)
print(f"ARNm normal:                 5'- {mrna_seq} -3'")
print(f"Péptido normal:                  {protein_to_stop} (Longitud: 3 aa)")
print(f"\nARNm con stop-loss (3'-UTR):  5'- {mut_stoploss_seq} -3'")
print(f"Péptido aberrante extendido:     {prot_mut_stoploss} (Longitud: {len(prot_mut_stoploss)} aa)")
print("   Secuencia extendida:          Met-Tyr-Ala-Gln-Ala (extensión C-terminal anómala)")

print("\n>> IMPACTO CELULAR Y PATOLÓGICO:")
print("1. Extensión C-terminal aberrante: Altera el plegamiento y puede causar toxicidad.")
print("2. Estancamiento ribosómico: Si no hay codón críptico, se traduce la cola poli-A (polilisina),")
print("   bloqueando el túnel ribosómico y activando la ruta de degradación Non-stop Decay (NSD).")
print("=" * 65)
```

### Paso 6 (Extra interactivo): Función modular para traducir y analizar cualquier ARNm

Encapsulamos la traducción en una función reutilizable `analizar_traduccion_arnm(secuencia_arnm)` que:
1. Valida que la secuencia esté en tripletes exactos (múltiplo de 3).
2. Comprueba si comienza con el codón de inicio canónico `AUG`.
3. Traduce identificando la presencia y posición del codón de parada.

```python
def analizar_traduccion_arnm(arnm_str: str) -> dict:
    """
    Traduce un transcrito de ARNm a proteína y genera un reporte bioinformático.
    """
    secuencia = arnm_str.strip().upper().replace("T", "U")  # Asegurar formato ARN
    
    if len(secuencia) % 3 != 0:
        print(f"[AVISO] La secuencia tiene {len(secuencia)} bases (no es múltiplo de 3; marco incompleto).")
    
    seq = Seq(secuencia)
    codones = [secuencia[i:i+3] for i in range(0, len(secuencia), 3)]
    peptido_con_stop = seq.translate()
    peptido_funcional = seq.translate(to_stop=True)
    
    tiene_inicio_canónico = codones[0] == "AUG" if codones else False
    tiene_stop = "*" in str(peptido_con_stop)
    
    print(f"\n>>> Análisis de traducción para ARNm: {seq}")
    print(f"    Codones:                   {codones}")
    print(f"    Inicio canónico (AUG):     {'SÍ' if tiene_inicio_canónico else 'NO'}")
    print(f"    Codón de parada detectado: {'SÍ' if tiene_stop else 'NO'}")
    print(f"    Péptido maduro:            {peptido_funcional} ({len(peptido_funcional)} aa)")
    
    return {
        "codones": codones,
        "peptido": peptido_funcional,
        "con_stop": peptido_con_stop
    }

# Prueba con el transcrito del ejercicio:
analizar_traduccion_arnm("AUGUAUGCUUAA")

# Prueba con otro transcrito de prueba:
analizar_traduccion_arnm("AUGUUUAAAUAG")
```


## 📌 Resumen de lo aprendido

- **`Seq.translate()`:** Traduce ARNm consultando el código genético estándar, marcando el stop con `*`.
- **`Seq.translate(to_stop=True)`:** Produce directamente el polipéptido maduro truncando en el codón de parada.
- **Mutaciones en el inicio ($AUG \to GUG$):** En eucariotas abolen casi completamente la traducción debido a la especificidad del complejo 43S; en procariotas permiten inicio alternativo con menor eficiencia.
- **Mutaciones Stop-loss:** Conducen a lectura continua en la región $3'\text{-UTR}$, extensiones anómalas de la proteína y estancamiento ribosómico en la cola de poliadenilación.
